```python
import json
from datasets import load_from_disk
```

```python
test = load_from_disk("../datasets/t5-datasets/test")
```

```python
print(test)
```

```python
test_clean_samples_minerva = {}
test_ocr_samples_minerva = {}

idx = 0
for sample in test:
    ocr_sample = sample['ocr']
    clean_sample = sample['clean']

    test_ocr_samples_minerva[idx] = ocr_sample
    test_clean_samples_minerva[idx] = clean_sample

    idx += 1
```

```python
# save the json files
with open('../INPUT/test_ocr_inputs_minerva3B-base.json', 'w', encoding='utf-8') as f:
    json.dump(test_ocr_samples_minerva, f, ensure_ascii=False, indent=4)

with open('../INPUT/test_clean_samples_minerva3B-base.json', 'w', encoding='utf-8') as f:
    json.dump(test_clean_samples_minerva, f, ensure_ascii=False, indent=4)
```
