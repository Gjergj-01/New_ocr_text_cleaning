# Plot results

```python
import matplotlib.pyplot as plt
import json
```

```python
mt5_scores_path = "../results/mt5-base/mt5-base_final_scores.json"
minerva3B_scores_path = "../results/minerva3B-base/minerva3B_final_scores.json"
```

### Read final scores

```python
with open(mt5_scores_path) as f:
    mt5_scores = json.load(f)

with open(minerva3B_scores_path) as f:
    minerva3B_scores = json.load(f)
```

```python
def bar_plot(data, filename):

    score_names = list(data.keys())
    values = list(data.values())

    colors = [
    '#E63946',  # Crimson Red
    '#F4A261',  # Coral Orange
    '#E76F51',  # Terracotta
    '#2A9D8F',  # Teal
    '#264653',  # Deep Blue-Green
    '#4A90E2',  # Sky Blue
    '#9B51E0',  # Amethyst Purple
    '#D946EF',  # Magenta
    '#FACC15',  # Bright Gold
    '#10B981',  # Emerald Green
    '#6B7280'   # Cool Slate Gray
    ]

    plt.figure(figsize=(10,6))

    bars = plt.bar(score_names, values, width=0.5, color=colors, edgecolor="black")
    for bar, score_name in zip(bars, values):
        bar.set_label(score_name)

    plt.xlabel("Score method")
    plt.ylabel("Values")
    plt.title("Scores")

    legend_labels = [f'{k} ({v:.4f})' for k, v in data.items()]
    plt.legend(bars, legend_labels, title="Score method", loc="upper left", bbox_to_anchor=(1.05, 1), borderaxespad=0)

    plt.ylim(0, 1)
    plt.xticks(rotation=45)

    plt.bar_label(bars, fmt='%.2f', padding=3)

    plt.tight_layout()

    plt.savefig(filename, dpi=300, bbox_inches='tight')

    plt.show()
```

```python
mt5_path = "../scores_plot/mt5_scores.png"
minerva3B_path = "../scores_plot/minerva3B_scores.png"

#bar_plot(mt5_scores, mt5_path)
bar_plot(minerva3B_scores, minerva3B_path)
```
